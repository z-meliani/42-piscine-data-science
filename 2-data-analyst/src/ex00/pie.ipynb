```python
import duckdb
import matplotlib.pyplot as plt
```

```python
project_root_path = "../../"
customers_path = project_root_path + "data/customers.parquet"
```

```python
# Load customers table
customers = duckdb.read_parquet(customers_path)
```

```python
# Count the nimber of records for each event type
pie = duckdb.sql("SELECT event_type, COUNT(*) AS event_count FROM customers GROUP BY event_type").fetchnumpy()
```

```python
fig, ax = plt.subplots()
ax.pie(pie['event_count'], labels=pie['event_type'], autopct='%1.1f%%')
ax.set_title('Distribution of event types')

plt.show()
```
