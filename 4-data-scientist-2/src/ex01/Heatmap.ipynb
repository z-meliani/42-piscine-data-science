```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("../../data/Train_knight.csv")
knight_code = {'Sith': 0, 'Jedi': 1}
df['knight'] = df['knight'].map(knight_code)
```

```python
plt.figure(figsize=(5, 4))
sns.heatmap(df.corr())
plt.show()
```
