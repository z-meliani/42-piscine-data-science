```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../../data/Train_knight.csv")
knight_code = {'Sith': 0, 'Jedi': 1}
df['knight'] = df['knight'].map(knight_code)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

pca = PCA()
pca.fit(X_scaled)

var_per = pca.explained_variance_ratio_ * 100
cum_var_per = np.cumsum(var_per)
print("Variances (Percentage):")
print(var_per)
print("\n")
print("Cumulative Variances (Percentage):")
print(cum_var_per)
```

```python
fig, ax = plt.subplots()
ax.plot(np.arange(1, cum_var_per.shape[0] + 1), cum_var_per)
ax.set_xlabel('Number of components')
ax.set_ylabel('Explained Variance (%)')
plt.show()
```
