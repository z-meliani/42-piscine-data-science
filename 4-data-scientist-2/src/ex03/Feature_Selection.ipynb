```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
df = pd.read_csv("../../data/Train_knight.csv")
knight_code = {'Sith': 0, 'Jedi': 1}
df['knight'] = df['knight'].map(knight_code)
```

```python
# Standardization
df_scaled = df.copy()
df_scaled = df_scaled.transform(lambda x: (x - x.mean()) / x.std(ddof=0))
```

```python
# Variance inflation factor
df_vif = pd.DataFrame(index=df_scaled.columns)

df_vif["VIF"] = pd.Series(range(df_scaled.shape[1]), index=df_scaled.columns).apply(
    lambda i: variance_inflation_factor(df_scaled.values, i)
    )

df_vif["Tolerance"] = 1 / df_vif["VIF"]
df_vif = df_vif[df_vif["VIF"] < 5].sort_values(by="VIF")
df_vif
```
