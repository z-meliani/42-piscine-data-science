```python
import pandas as pd
```

```python
train_knight = pd.read_csv('../../data/Train_knight.csv')

# Convert knight type to numerical value
knight_code = {'Sith': 0, 'Jedi': 1}
train_knight['knight'] = train_knight['knight'].map(knight_code)

# Compute the Pearson correlation coefficient
knight_corr = train_knight.corr(method='pearson')['knight'].sort_values(ascending=False)
knight_corr_index = knight_corr.abs().sort_values(ascending=False).index
knight_corr = knight_corr.loc[knight_corr_index]
knight_corr
```
