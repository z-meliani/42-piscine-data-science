```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
test_knight = pd.read_csv('../../data/Test_knight.csv')
train_knight = pd.read_csv('../../data/Train_knight.csv')
```

```python
norm_test_knight = (test_knight - test_knight.min()) / (test_knight.max() - test_knight.min())

knight_col = train_knight['knight']
train_knight = train_knight.loc[:, ~train_knight.columns.isin(['knight'])]
norm_train_knight = (train_knight - train_knight.min()) / (train_knight.max() - train_knight.min())
norm_train_knight['knight'] = knight_col
```

```python
fig, ax = plt.subplots(2, 1, figsize=(8, 8))

# Deflection / Push
(norm_train_knight[norm_train_knight["knight"] == "Jedi"]
 .plot.scatter(x="Push", y="Deflection", ax=ax[0], label="Jedi",
               c='blue', alpha=0.5))
(norm_train_knight[norm_train_knight["knight"] == "Sith"]
 .plot.scatter(x="Push", y="Deflection", ax=ax[0], label="Sith",
               c='red', alpha=0.5))

norm_test_knight.plot.scatter(x="Empowered", y="Stims", ax=ax[1],
                              label="Knight", c='green', alpha=0.5)

plt.show()
```
