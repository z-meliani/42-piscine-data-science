```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
test_knight = pd.read_csv('../../data/Test_knight.csv')
train_knight = pd.read_csv('../../data/Train_knight.csv')
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 10))

# Stims / Empowered
(train_knight[train_knight["knight"] == "Jedi"]
 .plot.scatter(x="Empowered", y="Stims", ax=ax[0, 0], label="Jedi",
               c='blue', alpha=0.5))
(train_knight[train_knight["knight"] == "Sith"]
 .plot.scatter(x="Empowered", y="Stims", ax=ax[0, 0], label="Sith",
               c='red', alpha=0.5))
test_knight.plot.scatter(x="Empowered", y="Stims", ax=ax[1, 0], label='Knight',
                         c='green', alpha=0.5)

# Deflection / Push
(train_knight[train_knight["knight"] == "Jedi"]
 .plot.scatter(x="Push", y="Deflection", ax=ax[0, 1], label="Jedi",
               c='blue', alpha=0.5))
(train_knight[train_knight["knight"] == "Sith"]
 .plot.scatter(x="Push", y="Deflection", ax=ax[0, 1], label="Sith",
               c='red', alpha=0.5))
test_knight.plot.scatter(x="Push", y="Deflection", ax=ax[1, 1], label='Knight',
                         c='green', alpha=0.5)

plt.show()
```
