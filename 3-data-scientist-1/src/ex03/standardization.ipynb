```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
test_knight = pd.read_csv('../../data/Test_knight.csv')
train_knight = pd.read_csv('../../data/Train_knight.csv')
```

```python
std_test_knight = (test_knight - test_knight.mean()) / test_knight.std()

knight_col = train_knight['knight']
train_knight = train_knight.loc[:, ~train_knight.columns.isin(['knight'])]
std_train_knight = (train_knight - train_knight.mean()) / train_knight.std()
std_train_knight['knight'] = knight_col
```

```python
fig, ax = plt.subplots(2, 1, figsize=(8, 8))

# Stims / Empowered
(std_train_knight[std_train_knight["knight"] == "Jedi"]
 .plot.scatter(x="Empowered", y="Stims", ax=ax[0], label="Jedi",
               c='blue', alpha=0.5))
(std_train_knight[std_train_knight["knight"] == "Sith"]
 .plot.scatter(x="Empowered", y="Stims", ax=ax[0], label="Sith",
               c='red', alpha=0.5))

std_test_knight.plot.scatter(x="Empowered", y="Stims", ax=ax[1],
                              label="Knight", c='green', alpha=0.5)

plt.show()
```
