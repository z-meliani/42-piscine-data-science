```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
test_knight = pd.read_csv('../../data/Test_knight.csv')

fig, ax = plt.subplots(6, 5, figsize=(20, 25))

for i, column in enumerate(test_knight.columns):
    row = i // 5
    col = i % 5
    ax[row, col].hist(test_knight[column], bins=40, color='green', alpha=0.6)
    ax[row, col].set_title(column)

plt.show()
plt.tight_layout()
```

```python
train_knight = pd.read_csv('../../data/Train_knight.csv')
sith_knight = train_knight[train_knight['knight'] == 'Sith'].drop(columns=['knight'])
jedi_knight = train_knight[train_knight['knight'] == 'Jedi'].drop(columns=['knight'])

fig, ax = plt.subplots(6, 5, figsize=(25, 25))
for i, column in enumerate(jedi_knight.columns):
    row = i // 5
    col = i % 5
    ax[row, col].hist(jedi_knight[column], bins=40, color='blue', alpha=0.5, label='Jedi')
    ax[row, col].hist(sith_knight[column], bins=40, color='red', alpha=0.5, label='Sith')
    ax[row, col].set_title(column)
    ax[row, col].legend(loc='upper right')

plt.show()
plt.tight_layout()
```
